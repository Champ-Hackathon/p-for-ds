**A. Write a Python program to create a scatter plot of Age and Height of 5 students.**

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([15,20,21,25,26])
y = np.array([180,170,145,156,160])

plt.scatter(x, y)
plt.xlabel("Age")
plt.ylabel("Height in cm")
plt.title("Scatter Plot Age and Height")
plt.show()
```

B. **Write a Python program to create a bar chart showing marks of 5 students.**


```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(["Nitesh", "Rahul", "Amit", "Kaushal","Sunny"])
y = np.array([80, 88, 78, 65,70])

plt.bar(x, y, color = "red")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Bar chart Student and Marks")
plt.show()
```

**C. Write a Python program to create a histogram showing the distribution of marks of 10 students.**

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([50, 80, 70, 75, 78, 82, 81, 83, 90, 65])

plt.hist(y)
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Histogram of Student Marks")
plt.show()
```

**D. Write a Python program to create a pie chart showing the number of students in different courses.**

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([35, 25, 25, 15])
mylabels = ["Bsc.IT", "Bsc.Biotech", "Bsc.CS", "BMMAC"]

plt.pie(y, labels = mylabels)
plt.title("Pie chart of Student Course")
plt.show()
```
