A. Pandas DataFrame filter() Method

```python
import pandas as pd

data = {
  "name": ["Nitesh", "Amit", "Rahul"],
  "marks": [85, 80, 75],
  "qualified": [True, False, False]
}
df = pd.DataFrame(data)

newdf = df.filter(items=["name", "marks"])
print(newdf)
```

B. Python statistics Module

```python
import statistics

print(statistics.mean([1, 9, 4, 7, 10, 11, 11]))
print(statistics.mean([1, 3, 5, 8, 4, 11]))
print(statistics.mean([-8, 5.5, -3.4, 8.0, -7, 12]))
```

```python
import statistics

print(statistics.median([1, 9, 4, 7, 10, 11, 11]))
print(statistics.median([1, 3, 5, 8, 4, 11]))
print(statistics.median([-8, 5.5, -3.4, 8.0, -7, 12]))
```

```python
import statistics

print(statistics.mode([1, 7, 4, 7, 10, 11, 11]))
print(statistics.mode([1, 4, 5, 8, 4, 11]))
print(statistics.mode([-8, 5.5, -3.4, 8.0, 5.5, 12]))
```

```python
import statistics

print(statistics.stdev([1, 9, 4, 7, 10, 11, 11]))
print(statistics.stdev([1, 3, 5, 8, 4, 11]))
print(statistics.stdev([-8, 5.5, -3.4, 8.0, -7, 12]))
```

```python
import statistics

print(statistics.variance([1, 9, 4, 7, 10, 11, 11]))
print(statistics.variance([1, 3, 5, 8, 4, 11]))
print(statistics.variance([-8, 5.5, -3.4, 8.0, -7, 12]))
```

C. Dictionary keys are given in sorted order.

```python
import pandas as pd

dictionary = {'A': 140, 'B': 45, 'C': 78}

series = pd.Series(dictionary)

print(series)
```

```python
import pandas as pd

dictionary = {'B': 140, 'C': 45, 'A': 78}

series = pd.Series(dictionary)

print(series)
```

Indexing

```python
import pandas as pd

dictionary = {'A': 50, 'B': 10, 'C': 80}

series = pd.Series(dictionary, index=['B', 'C', 'A'])

print(series)
```

D. Boolean Indexing in Pandas

```python
import pandas as pd
 
dict = {'name':["Nitesh", "Pankaj", "Amit", "Rahul"],
        'degree': ["MBA", "BCA", "M.Tech", "MBA"],
        'score':[90, 40, 80, 98]}
 
df = pd.DataFrame(dict, index = [True, False, True, False])
 
print(df)
```
